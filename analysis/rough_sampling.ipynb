```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
A_min = 2
A_max = 5
B_A_min = 1.1
B_A_max = 2
n_A = 5
n_B = 7
n_Du = 2
n_Dv = 7
```

```python
A_vals = np.linspace(A_min, A_max, n_A)
B_A_vals = np.linspace(B_A_min, B_A_max, n_B)
Du_vals = np.array([1,2])
D_mult_vals = np.linspace(7, 15, n_Dv)
```

```python
vals = []
for A in A_vals:
    for B_A in B_A_vals:
        for Du in Du_vals:
            for D_mult in D_mult_vals:
                vals.append([A, A * B_A, Du, Du * D_mult])

df = pd.DataFrame(vals, columns=["A", "B", "Du", "Dv"])
df.to_csv("../data/rough2.csv", index=False)
```
