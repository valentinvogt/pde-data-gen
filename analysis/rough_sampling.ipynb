```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# bruss
A_min = 4.5
A_max = 5.5
B_A_min = 1.8
B_A_max = 2.2
n_A = 10
n_B = 10
n_Du = 1
n_Dv = 1
ratio_D_min = 7
ratio_D_max = 7
```

```python
n = n_A * n_B * n_Du * n_Dv * 2
print(n, n * 4 * 1/(60**2))
```

```python
A_vals = np.linspace(A_min, A_max, n_A)
B_A_vals = np.linspace(B_A_min, B_A_max, n_B)
Du_vals = np.array([1.0])
D_mult_vals = np.linspace(ratio_D_min, ratio_D_max, n_Dv)
vals = []
for A in A_vals:
    for B_A in B_A_vals:
        for Du in Du_vals:
            for D_mult in D_mult_vals:
                # if Du * D_mult > 24:
                #     continue
                # if A * B_A > 10:
                #     continue
                vals.append([A, A * B_A, Du, Du * D_mult])

df = pd.DataFrame(vals, columns=["A", "B", "Du", "Dv"])
```
