```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
A_min = 0.03
A_max = 0.05
B_A_min = 0.8
B_A_max = 2.5
n_A = 10
n_B = 10
n_Du = 1
n_Dv = 10
ratio_D_min = 0.2
ratio_D_max = 1.2
```

```python
n = n_A * n_B * n_Du * n_Dv * 2
print(n, n * 19 *1/(60**2))
```

```python
A_vals = np.linspace(A_min, A_max, n_A)
B_A_vals = np.linspace(B_A_min, B_A_max, n_B)
Du_vals = np.array([0.1])
D_mult_vals = np.linspace(ratio_D_min, ratio_D_max, n_Dv)
```

```python
vals = []
for A in A_vals:
    for B_A in B_A_vals:
        for Du in Du_vals:
            for D_mult in D_mult_vals:
                # if Du * D_mult > 24:
                #     continue
                # if A * B_A > 10:
                #     continue
                vals.append([A, A * B_A, Du, Du * D_mult])

df = pd.DataFrame(vals, columns=["A", "B", "Du", "Dv"])
```

```python
df.to_csv('../data/gs_rough.csv')
```
