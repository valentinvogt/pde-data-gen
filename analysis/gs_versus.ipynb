```python
model = "gray_scott"
ds_id = "new_ball"
ds, output_dir = get_dataset(model, ds_id, "SCRATCHDIR")
df = ds.df
df["ratio_b_a"] = df["B"] / df["A"]
df["ratio_d"] = df["Dv"] / df["Du"]
df_ball = df.copy()
```

```python
fig1 = px.scatter_3d(
    df_rough,
    x="A",
    y="ratio_b_a",
    z="rel_std_u", 
    color_continuous_scale=px.colors.sequential.Bluered,
)
fig1.update_traces(marker_size=5)

# Extract the trace(s) from the first figure
fig = go.Figure(data=fig1.data)

# Second scatter (manual trace)
fig.add_trace(
    go.Scatter3d(
        x=df_ball["A"],
        y=df_ball["ratio_b_a"],
        z=df_ball["rel_std_u"],
        mode='markers',
        marker=dict(size=4, color='black'),
        name='df_final points'
    )
)

fig.show()
```
