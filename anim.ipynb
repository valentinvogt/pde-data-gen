```python
import numpy as np
import netCDF4 as nc
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file = nc.Dataset("gathered_data.nc", "r")
```

```python
u = np.array(file["data"][0][0][0])
```

```python
fig = plt.figure(figsize=(7, 7))
ax = fig.add_axes([0.1, 0.1, 0.75, 0.75])  # main plot
im = ax.imshow(
    u.T,
    origin="lower",
    extent=(0.2, 5, 0.5, 3),
    aspect="auto",
    cmap="viridis",
)
ax.set_xlabel("$A$")
ax.set_ylabel("$B/A$")
plt.title("$u$ as $t\\rightarrow\\infty$ in parameter space", fontdict={"fontsize": 16})
# colorbar in separate axes
cax = fig.add_axes([0.88, 0.1, 0.04, 0.75])
fig.colorbar(im, cax=cax, label="$u$")
fig.savefig("param_map.png", dpi=300)
```
